```python
merged_df = pd.merge(pd.Series(pd.date_range(start=df["Date"].min(), end=df["Date"].max()), name="Date"), df, on="Date", how="left")
merged_df.head()
```

```python
trans_by_date = merged_df["Date"].value_counts()
all_time = trans_by_date[(trans_by_date.index >= pd.Timestamp(2018,7,1)) & (trans_by_date.index < pd.Timestamp(2019,6,30))].sort_index()
all_time.index = all_time.index.strftime('%d')
ax = all_time.plot(figsize=(15,3))
ax.set_xticks(np.arange(len(all_time)))
ax.set_xticklabels(all_time.index)
plt.title("All Time Sales")
plt.savefig("All Time Sales.png", bbox_inches="tight")
plt.show()
```

Insight:  Transaction is high a period and suddenly falls in another period say month

```python
# Count transactions by date
trans_by_date = merged_df["Date"].value_counts()

# Filter by date range
all_time = trans_by_date[
    (trans_by_date.index >= pd.Timestamp(2018,7,1)) &
    (trans_by_date.index < pd.Timestamp(2019,6,30))
].sort_index()

# Group by month
monthly = all_time.groupby(all_time.index.to_period("M")).sum()

# Format index as YYYY-MM for x-axis
monthly.index = monthly.index.strftime('%Y-%m')

# Plot
ax = monthly.plot(figsize=(15,3))
ax.set_xticks(np.arange(len(monthly)))
ax.set_xticklabels(monthly.index, rotation=45)
plt.title("Monthly Sales (Jul 2018 – Jun 2019)")
plt.savefig("Monthly Sales.png", bbox_inches="tight")
plt.show()
```

```python
trans_by_date = merged_df["Date"].value_counts()
dec = trans_by_date[(trans_by_date.index >= pd.Timestamp(2018,12,1)) & (trans_by_date.index < pd.Timestamp(2019,1,1))].sort_index()
dec.index = dec.index.strftime('%d')
ax = dec.plot(figsize=(15,3))
ax.set_xticks(np.arange(len(dec)))
ax.set_xticklabels(dec.index)
plt.title("December Sales")
plt.savefig("December Sales.png", bbox_inches="tight")
plt.show()
```

Insight:  Sales are increasing towards Dec 24 and had a sudden fall on Dec 25 which may be because the shop is closed due to Christmas.

```python
df["Brand"] = df["Product"].str.split().str[0]
df.head()
```

```python
def clear_brand_names(line) :
    brand = line["Brand"]
    if brand == "Dorito" :
        return "Doritos"
    elif brand == "GrnWves" or brand == "Grain" :
        return "Grain Waves"
    elif brand == "Infzns" :
        return "Infuzions"
    elif brand == "Natural" or brand == "NCC" :
        return "Natural Chip Co"
    elif brand == "Red" :
        return "RRD"
    elif brand == "Smith" :
        return "Smiths"
    elif brand == "Snbts" :
        return "Sunbites"
    elif brand == "WW" :
        return "Woolworths"
    else :
        return brand
```

```python
df["Brand"] = df.apply(lambda line : clear_brand_names(line), axis=1)
df.head()
```

```python
df["Brand"].value_counts(ascending=True).plot.barh(figsize = (10, 5))
plt.show()
```

```python
mainstream = df["Subscription"] == "Mainstream"
budget_premium = (df["Subscription"] == "Budget") | (df["Subscription"] == "Premium")
young_midage = (df["Group"] == "YOUNG SINGLES/COUPLES") | (df["Group"] == "MIDAGE SINGLES/COUPLES")

a = df[young_midage & mainstream]["Sales"]
b = df[young_midage & budget_premium]["Sales"]

stat, pval = ttest_ind(a.values, b.values, equal_var=False)
print(pval)
pval < 0.0000001
```
